# Introducción a las listas

Una lista es una estructura de datos que tiene las siguients características:  

*   Es heterogénea ya que puede contener datos de diferente tipo: cadena (string), enteros (int), flotante (float), así como listas y otras estructuras de datos.
*   Es mutable debido a que podemos cambiar o modificar los elementos que contiene.

Una lista se crea escribiendo cada elemento dentro de corchetes **[ ]** separados por comas.


````python
#Crear una lista
L = [1,2,3]
# Imprimir la lista
print(L)
````

También se puede obtener una parte de una lista especificando el índice inicial y el índice final. Este último debe ser un número mayor al último elemento que se desea.

````python
#Crear la lista
L = [10,15,5,2,20,8]
#Imprimir del elemento 3 al 6
print(L[2:6])
````

Para conocer el tamaño de una lista se puede utilizar la función len(), la cual devuelve la cantidad de objetos que contiene.

````python
#Imprimir el tamaño de la lista
len(L)
````

Es posible concatenar dos listas utilizando el operador +.

````python
L1 = [1,2,3]
L2 = [4,5]
L3 = L1 + L2
print(L3)
````

# Iterar listas

Se puede acceder a cada elemento de una lista utilizando el ciclo for.

````python
Lista = [10,2,8,9]
for l in Lista:
  print(l)
````

Para obtener el índice del elemento se puede utilizar la función enumerate():

````python
for i,l in enumerate(Lista):
  print(i,l)
````

También se puede iterar utilizando len()

````python
for i in range(0, len(Lista)):
    print(Lista[i])
````

# Métodos

Las listas integran unos métodos que se pueden utilizar para realizar algunas acciones. A continuación se enlistan algunos de ellos:


1.   **append():** Agrega un elemento al final de la lista.
2.   **count():** Recibe un elemento y devuelve un número correspondiente a la cantidad de veces que aparece ese elemento en la lista.
3. **index():** Recibe un elemento  devuelve el índice de la posición del elemento en la lista.
4. **insert():** Agrega un elemento en la posición indicada
5. **sort():** Ordena los elementos de la lista
6. **copy():** Devuelve una nueva lista que contiene una copia de la original.



````python
#Crear la lista
L = [10,15,5,2,20,8]
#Utilizar append() para agregar un elemento al final de la lista
L.append(1)
print(L)
````

````python
#Utilizar count() para conocer la cantidad de veces que aparece el número 15 en la lista
n = L.count(5)
print(n)
````

````python
#Utilizar index() para conocer la posición del número 5 en la lista
i = L.index(15)
print(i)
````

````python
#Utilizar insert() para añadir el número 11 en la posición 4
L.insert(4,11)
print(L)
````

````python
#Utilizar sort() para ordenar los elementos de la lista
#Orden ascendente
L.sort()
print(L)
#Orden descendente
L.sort(reverse=True)
print(L)
````

````python
#Copiar una lista
copia = L[:]
print(copia)
#Copiar una lista con el método copy()
copia = L.copy()
print(copia)
````

# Listas de dos dimensiones

Las listas pueden contener otras listas como elementos. Esto es muy útil para representar tablas de valores que contienen información organizada en filas y columnas. En este caso, cada elemento de la lista se puede identificar por medio de dos índices donde el primero corresponde a la fila y el segundo a la columna.

Considera la siguiente lista L:

````python
L = [[10,8,7,10],[8,9,5,10],[5,6,8,7]]
````

En la imagen se muestra la lista L con sus respectivas filas y columnas. Podemos acceder a cada uno de los elementos a través de la fila y la columna de la siguiente manera:  
`L[i][j]` donde i es la fila y j la columna.

listas_tabla.PNG

````python
#Imprimir el valor de la fila 0 y columna 2  = 7
valor = L[0][2]
print(valor)
````

Para acceder uno a uno a los elementos de la lista, se puede utilizar dos ciclos for anidados:

````python
#Imprimir los elementos de la lista
#El primer ciclo for obtiene las filas, una a la vez.
#En cada iteración del primer ciclo for, el for interno obtiene el elemento de cada columna de la fila.
for fila in L:
  for num in fila:
    #Imprime el elemento con un espacio
    print(num, end=' ')
  #Imprime salto de línea
  print()
````

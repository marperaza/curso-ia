# Series

Una serie es como un *numpy array* de una dimensión donde cada elemento tiene un índice. Esta estructura es parecida a los diccionarios. Las series soportan indexado personalizado incluyendo índices no numéricos como cadenas.

Una característica especial, es que las series pueden contener datos faltantes que se ignoran por defecto.

De forma predeterminada, una serie tiene índices de números enteros numerados secuencialmente desde 0 y se puede crear con la función Series() a partir de una lista, un diccionario o un array.


````python
#Crear una serie a partir de una lista de números
S1 = pd.Series([90,100,85])
````

Pandas imprime una Serie en un formato de dos columnas con los índices en la columna de la izquierda y los valores en la columna de la derecha. Al final muestra el tipo de datos que contiene.

````python
print(S1)
````

Se pueden crear series con índices específicos pasando los índices al argumento index de la función Series().

````python
#Crear una serie a partir de una lista de números con índices de tipo cadena
S2 = pd.Series([90,100,85], index=['Ana','María','Carlos'])
print(S2)
````

Al crear una serie a partir de un diccionario, los índices de la Serie son las claves.

````python
#Crear una lista a partir de un diccionario
S3 = pd.Series({'Ana':90,'María':100,'Carlos':85})
print(S3)
````

Se puede crear una serie con una lista de elementos de tipo string.

````python
#Crear una lista
S4 = pd.Series(['Ana','María','Carlos'])
print(S4)
````

**Indexación**

Se puede acceder a un elemento de la serie indicando el índice entre corchetes **[ ]**.

````python
#Imprimir el primer elemento
e = S1[0]
print(e)
````

Para acceder algún elemento de la Serie con índices, por ejemplo, de tipo cadena se debe especificar el índice entre corchetes **[ ]**.

````python
#Acceder a un elemento
print(S3['María'])
````

**Métodos**

Con las series podemos utilizar varios métodos para hacer cálculos de estadística descriptiva.

Por ejemplo:

````python
#Cantidad de elementos con el método count()
print(S1.count())
#Promedio de los elementos con el método mean()
print(S1.mean())
#Menor elemento con el método min()
print(S1.min())
#Mayor elemento con el método max()
print(S1.max())
#Desviación estándar con el método std()
print(S1.std())
````

También es posible obtener todos estos cálculos y otros más con el método **describe()**.

````python
S1.describe()
````

**Atributos**

*   dtype: Devuelve el tipo de datos de los elementos que contiene.
*   values: Devuelve el array de valores.



````python
print(S1.dtype)
print(S1.values)
````

# Dataframe

Un dataframe es como un numpy array de dos dimensiones, compuesto de filas y columnas. Cada columna en un dataframe es una serie. Las series pueden estar conformadas con datos de diferente tipo (uno para cada columna).

Se puede crear un dataframe a partir de un diccionario utilizando la función **DataFrame()**.

````python
import pandas as pd
diccionario = {'Ana': [80,95,92],'María': [83,85,98],'Carlos': [75,81,96],'Rodrigo': [65,83,77],'Marcos': [91,89,76]}
D1 = pd.DataFrame(diccionario)
````

Pandas imprime un Dataframe de manera tabular con los índices de las filas a la izquierda y las claves o nombres de las columnas en la parte superior.

````python
D1
````

Como en las series, se pueden especificar los índices del dataframe con el atributo *index*.


````python
D1.index = ['Examen1','Examen2','Examen3']
D1
````

**Número de filas y columnas**

````python
D1.shape # retorna un Tuple asi: (filas, col)
````

**Información general de los datos**

````python
# Indica el numero de filas del dataset
# Muestra el numero de datos No Nulos por columna (valores validos)
# Tipo de dato de cada columna
# Tamaño total del dataset
D1.info()
````

**Tipos de datos**

````python
# Tipos de datos que existen en las columnas del dataframe
D1.dtypes
````

**Indexación**

Se puede obtener una columna específica escribiendo el nombre de la columna entre corchetes [ ].

````python
D1['Ana']
````

Para obtener filas específicas se pueden utilizar los atributos loc y iloc.

**loc** permite obtener una fila dada su etiqueta:

````python
D1.loc['Examen1']
````

**iloc** permite obtener las filas con los ínidices (números enteros):

````python
D1.iloc[1]
````

Ambos, loc y iloc, permiten extraer un rango de filas:

````python
#Obtener las filas 0 a 1 con loc
D1.loc['Examen1':'Examen2']
````

````python
#Obtener las filas 0 a 1 con iloc. El último índice no se considera.
D1.iloc[0:2]
````

Se pueden seleccionar filas específicas indicando los índices con una lista.

````python
#Obtener las filas 0 y 2 con loc
D1.loc[['Examen1','Examen3']]
````

````python
#Obtener las filas 0 y 2 con iloc.
D1.iloc[[0,2]]
````

Además, es posible seleccionar un subconjunto de datos seleccionando las filas y columnas del dataframe.

````python
#Filas 0 a 1 y columnas 1 a 3
D1.iloc[0:2,1:4]
````

````python
#Examen1 y Examen2 de las columnas Ana y Rodrigo
D1.loc['Examen1':'Examen2',['Ana','Rodrigo']]
````

````python
#Examen1 y Examen2 de las columnas Ana y Rodrigo
D1.iloc[0:2,[0,3]]
````

**Métodos**

El método **describe()** obtiene datos de estadística descriptiva por columna que incluyen la media, desviación estándar, máximo y mínimo.

````python
D1.describe()
````

El método **mean()** calcula la media de cada columna.

````python
D1.mean()
````

El método **head()** imprime las primeras 5 filas del dataframe. Se puede especificar la cantidad colocando el número dentro del paréntesis.

````python
D1.head(2)
````

El método **drop()** elimina una columna/fila del dataframe. Se debe especificar axis (0 para fila, 1 para columna).

````python
D1.drop('Ana', axis=1, inplace =True)
````

````python
D1
````

````python
D1.drop('Examen2',axis=0)
````
